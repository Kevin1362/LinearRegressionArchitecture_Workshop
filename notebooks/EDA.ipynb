# EDA (Robot Current Data)
This notebook quickly checks the robot current export used in the Linear Regression Architecture Workshop.

```
import pandas as pd
import matplotlib.pyplot as plt

from src.data_loader import load_from_csv, make_multi_robot
from src.preprocessing import clean_robot_current_df

csv_path = 'data/raw/RMBR4-2_export_test.csv'
df = load_from_csv(csv_path)
print(df.shape)
df.head()
```

```
df = make_multi_robot(df)
df = clean_robot_current_df(df)
print(df[['robot_id','Time']].head())
df.describe(include='all').T.head(20)
```

```
axis_col='Axis #1'
plt.figure()
for rid, g in df.groupby('robot_id'):
    plt.plot(g['Time'].iloc[:500], g[axis_col].iloc[:500], label=rid)
plt.title(f'{axis_col} (first 500 points)')
plt.xlabel('Time')
plt.ylabel('Current (A)')
plt.legend()
plt.show()
```
