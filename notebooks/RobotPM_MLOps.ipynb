# RobotPM_MLOps.ipynb
This notebook documents how the original **DataStreamVisualization_Workshop** project was extended into the **Linear Regression Architecture Workshop** project.

## Goal
- Move from **streaming + visualization** to a **reproducible ML experiment** that can be re-run from a config file.
- Add modular code for data loading, preprocessing, modeling, and evaluation.
- Track results for each experiment run.

---


## Recommended Additions (implemented)
1. **Folder structure** for an ML project (`data/`, `src/`, `configs/`, `experiments/`, `notebooks/`).
2. **Configuration file** (`configs/experiment_config.yaml`) with data source + hyperparameters.
3. **Experiment tracking** (`experiments/results.csv`) storing RMSE/MAE/R² per run.
4. **Modular pipeline** (`src/run_experiment.py`) to run end-to-end from the command line.
